```python
from dotenv import load_dotenv
import os

load_dotenv(override=True)

api_key = os.getenv("OPENAI_API_KEY")
default_model = os.getenv("OPENAI_DEFAULT_MODEL", 'gpt-6-luna')

print(api_key)
print(default_model)
```

```python
from openai import OpenAI

client = OpenAI()
```

```python
for i in client.models.list():
    print(i.id)
```

```python
res = client.responses.create(
    model=default_model,
    input="파이썬 리스트를 초보자도 쉽게 이해할 수 있게 두 문장으로 만들어줘"
)
print(res)
```

```python
reponse = client.responses.create(
    model=default_model,
    instructions='너는 자바 개발자야 스프링부트 프레임워크로 개발하고 있어',
    input='봄에 대해 알려줘'

)
print(reponse.output_text)
```

```python
from pydantic import BaseModel, Field

class Result(BaseModel):
    desc : str = Field(description="긍정, 부정, 중립 중 하나")
    confidense : int = Field(ge=0, le=100)
    reason : str
```

```python
aaa = Result(desc='긍정', confidense=90, reason='좋은 성과를 냈습니다.')
print(aaa.model_dump_json(indent=4))
```

```python
response = client.responses.create(
    model=default_model,
    tools=[{"type": "web_search"}],
    input="오늘 대한민국에서 발표된 AI 관련 주요 뉴스 2개를 출처와 함께 요약해 줘. 출처는 다음 줄에 출력해 줘.",
)
print(response.output_text)
```

```python
from dotenv import load_dotenv
import os

load_dotenv()
api_key = os.getenv('OPENAI_API_KEY')
default_model = os.getenv('OPENAI_DEFAULT_MODEL')
embedding_model = os.getenv('EMBEDDING_MODEL')

# print(api_key)
# print(default_model)
# print(embedding_model)

from openai import OpenAI
client = OpenAI()

prev_id = None

while (s:=(input("input: (quit=q)")).lower()) != 'q':
    response = client.responses.create(
    model=default_model,
    input= s,
    reasoning={"effort": "low"},
    previous_response_id=prev_id,
    )
    prev_id = response.id
    print(f"input: {s}")
    print(f"output: {response.output_text}\n")
```
